## 0. Setup


```python
import os
import sys
from pathlib import Path
import runpy

current_dir = Path.cwd()
project_root = current_dir.parent if current_dir.name == "notebooks" else current_dir
matplotlib_cache_dir = project_root / ".matplotlib_cache"
matplotlib_cache_dir.mkdir(parents=True, exist_ok=True)
os.environ["MPLCONFIGDIR"] = str(matplotlib_cache_dir)

import matplotlib
if "ipykernel" not in sys.modules:
    matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pandas as pd

pd.set_option("display.max_columns", 120)
pd.set_option("display.width", 160)
pd.options.display.float_format = "{:,.4f}".format

reports_dir = project_root / "reports"
figures_dir = reports_dir / "figures"
script_path = project_root / "src" / "seller_logistics_lift_analysis.py"

def show_image(path, width=12):
    image = plt.imread(path)
    height = width * image.shape[0] / image.shape[1]
    fig, ax = plt.subplots(figsize=(width, height))
    ax.imshow(image)
    ax.axis("off")
    plt.tight_layout()
    plt.show()

project_root
```

## 2. Seller Performance Scorecard

The scorecard segments sellers by commercial value and customer experience risk.

This creates a practical seller performance view:

- High value / high risk: prioritize operational intervention
- High value / standard risk: protect and grow
- Standard value / high risk: monitor or improve
- Low volume: avoid over-interpreting noisy rates
