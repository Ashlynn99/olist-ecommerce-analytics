# 09 Seller Monthly Risk Monitoring

This stage converts the static seller scorecard into a recurring monthly monitoring workflow.

The objective is to identify sellers whose customer-experience risk is high or deteriorating, while accounting for monthly sample size and commercial exposure.

## Monitoring components

- Seller-month performance panel
- Low-review, late-delivery, and cancellation risk
- Empirical-Bayes-style rate smoothing to reduce small-sample alerts
- Deterioration relative to each seller's prior history
- Commercial exposure and operational priority score
- Monthly watchlist and risk-status transitions
