# 10 Intervention Cost and Expected-Value Simulation

This stage translates the purchase-time and post-delivery risk rankings into business decision scenarios.

The analysis estimates how intervention coverage, contact cost, assumed effectiveness, and recovery value affect expected net value. It is a retrospective scenario simulation, not a measured causal impact study.

## Simulation components

- Base-case purchase-time prevention and post-delivery recovery strategies
- Expected net value and ROI across risk-coverage levels
- Break-even intervention effectiveness
- Incremental value versus randomly selecting the same number of orders
- Conservative, base, and optimistic scenarios
- Cost and effectiveness sensitivity analysis
