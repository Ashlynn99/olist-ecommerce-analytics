# 07 Purchase-Time Low-Review Risk Model

This stage builds an early risk-ranking model using only information available at or before purchase time.

The main methodological requirement is leakage control:

- Current-order delivery outcomes are excluded.
- Current-order review information is excluded.
- Seller-history features use only delivery or review events observed before the current purchase timestamp.
- Model performance is evaluated on a later time period.
