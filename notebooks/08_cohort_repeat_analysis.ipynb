# 08 Cohort and Observed Repeat-Purchase Analysis

This stage adds a customer-lifecycle view using `customer_unique_id` and delivered orders.

Because the dataset covers a limited historical window, the analysis uses the term **observed repeat behavior** rather than claiming true long-term retention. A fixed 90-day repeat window is used to reduce right-censoring.

## Analysis components

- Monthly first-purchase cohorts and observed repeat activity
- New-customer orders versus repeat orders
- 90-day repeat-purchase rate
- Days to second delivered order
- Descriptive relationship between first-order experience and repeat behavior
- Repeat behavior by first-purchase category
